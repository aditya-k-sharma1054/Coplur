```python
# C S V - F I L E
import csv
d = [
    ['Name', 'email', 'mobile'],
    ['ABC', 'email@som.com', '998877665543'],
    ['XYZ', 'qwsa@saa.com', '534522522523'],
    ['ABC2', 'email@som.com', '998877665543'],
    ['XYZ2', 'qwsa@saa.com', '534522522523'],    
    ['ABC3', 'email@som.com', '998877665543'],
    ['XYZ3', 'qwsa@saa.com', '534522522523'],
]
with open('data.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    #writer.writerows(d)
    for row in d:
        writer.writerow(row)
    
with open('data.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# E X C E P T I O N - H A N D L I N G :-
try:
    n = int(input("Enter a number: "))
    res = list(range(2, 3))
    res = 35 / n
    print(res)
except ZeroDivisionError:
    print("You cannot divide by zero.")
except SyntaxError:
    print("Syntax error occurred.")
except ValueError:
    print("Invalid input. Please enter a valid number.")
except TypeError:
    print("Invalid input type. Please enter a number.")
except NameError:
    print("Name isn't defined.")
except IndexError:
    print("Index out of range.")
except ValueError:
    print("Value error occurred.")
except KeyError:
    print("Key error occurred.")
except IndentationError:
    print("Indentation error occurred.")
```

```python
# D A T A B A S E - H A N D L I N G :-
import sqlite3
conn = sqlite3.connect('dbb.db')
sql = '''
CREATE TABLE IF NOT EXISTS employee
(
    id INTEGER PRIMARY KEY autoincrement,
    name VARCHAR(50),
    mobile VARCHAR(20),
    city VARCHAR(50)
)
'''
conn.execute(sql)
conn.commit()
conn.close()
```
