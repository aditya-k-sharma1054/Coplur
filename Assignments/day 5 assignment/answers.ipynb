```python
# Creating a CSV file for address book with name, address, mobile and email

import csv

address_data = []

for i in range(3):
    print("\nEnter details for record", i + 1)

    name = input("Enter Name: ")
    address = input("Enter Address: ")
    mobile = input("Enter Mobile: ")
    email = input("Enter Email: ")

    address_data.append([name, address, mobile, email])

with open("address_book.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Address", "Mobile", "Email"])
    writer.writerows(address_data)

print("\nCSV file created successfully as address_book.csv")
```

```python
# Creating and connecting to SQLite database

import sqlite3

conn = sqlite3.connect("assignment5.db")
cursor = conn.cursor()

print("Database created and connected successfully")
```

```python
# Creating required tables in the database

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS courses (
    id INTEGER PRIMARY KEY,
    course_name TEXT,
    duration TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS teachers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    subject TEXT
)
""")

conn.commit()

print("Tables created successfully")
```

```python
# Inserting sample records into tables

cursor.execute("INSERT INTO students (name, age) VALUES ('Aarav Mehta', 20)")
cursor.execute("INSERT INTO students (name, age) VALUES ('Diya Sharma', 22)")
cursor.execute("INSERT INTO students (name, age) VALUES ('Rohan Gupta', 19)")
cursor.execute("INSERT INTO students (name, age) VALUES ('Sneha Patel', 21)")

cursor.execute("INSERT INTO courses (course_name, duration) VALUES ('Python Programming', '3 Months')")
cursor.execute("INSERT INTO courses (course_name, duration) VALUES ('Database Systems', '2 Months')")
cursor.execute("INSERT INTO courses (course_name, duration) VALUES ('Data Science', '4 Months')")
cursor.execute("INSERT INTO courses (course_name, duration) VALUES ('Web Development', '3 Months')")

cursor.execute("INSERT INTO teachers (name, subject) VALUES ('Mr. Karan Singh', 'Python')")
cursor.execute("INSERT INTO teachers (name, subject) VALUES ('Ms. Priya Nair', 'DBMS')")
cursor.execute("INSERT INTO teachers (name, subject) VALUES ('Dr. Anjali Desai', 'Data Science')")
cursor.execute("INSERT INTO teachers (name, subject) VALUES ('Mr. Rahul Verma', 'Web Development')")
conn.commit()

print("Records inserted successfully")
```

```python
# Performing different SELECT operations on database tables

print("Students Table:")
print(cursor.execute("SELECT * FROM students").fetchall())

print("\nCourses Table:")
print(cursor.execute("SELECT * FROM courses").fetchall())

print("\nTeachers Table:")
print(cursor.execute("SELECT * FROM teachers").fetchall())

print("\nStudents with age > 20:")
print(cursor.execute("SELECT name FROM students WHERE age > 20").fetchall())
```

```python
# Updating existing records in the database

cursor.execute("UPDATE students SET age = 21 WHERE name = 'Aarav Mehta'")
conn.commit()

print("Updated Students Table:")
print(cursor.execute("SELECT * FROM students").fetchall())
```

```python
# Deleting specific records from the database

cursor.execute("DELETE FROM students WHERE name = 'Diya Sharma'")
conn.commit()

print("Final Students Table after deletion:")
print(cursor.execute("SELECT * FROM students").fetchall())

conn.close()
```
