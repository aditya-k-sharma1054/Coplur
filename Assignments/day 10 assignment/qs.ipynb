```python
import numpy as np
a = np.array([[6, -8, 73, -110], [np.nan, -8, 0, 94]])
a = np.nan_to_num(a, nan=0)
r = a[[1, 0]]
c = a[:, [3, 2, 1, 0]]
print(r)
print(c)
```

```python
import numpy as np
a = np.arange(24).reshape(2, 3, 4)
b = np.moveaxis(a, 0, 2)
print(b)
```

```python
import numpy as np
a = np.array([
    [4, np.nan, 7],
    [2, 5, np.nan],
    [6, 8, 9]
])
m = np.nanmean(a, axis=0)
i = np.where(np.isnan(a))
a[i] = np.take(m, i[1])
print(a)
```

```python
import numpy as np
a = np.array([4, -3, 7, -1, 0, -9])
a = np.where(a < 0, 0, a)
print(a)
```
