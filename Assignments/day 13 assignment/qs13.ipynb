```python
import numpy as np
arr=np.array([[6,-8,73,-110],[np.nan,-8,0,94]])
arr=np.nan_to_num(arr,nan=0)
print(arr)
print(arr.T)
```

```python
arr=np.arange(24).reshape(2,3,4)
new_arr=np.moveaxis(arr,[0,1,2],[2,0,1])
print(arr.shape)
print(new_arr.shape)
print(new_arr)
```

```python
arr=np.array([[10,np.nan,30],[40,50,np.nan],[70,80,90]])
col_mean=np.nanmean(arr,axis=0)
inds=np.where(np.isnan(arr))
arr[inds]=np.take(col_mean,inds[1])
print(arr)
```

```python
arr=np.array([[6,-8,73,-110],[20,-5,15,-2]])
arr=np.where(arr<0,0,arr)
print(arr)
```

```python
arr1=np.array([[3,4],[6,8]])
arr2=np.array([[1,0],[5,7]])
avg=(arr1+arr2)/2
combined=np.concatenate((arr1.flatten(),arr2.flatten()))
values,counts=np.unique(combined,return_counts=True)
mode=values[np.argmax(counts)]
print("Average Array")
print(avg)
print("Mean =",np.mean(combined))
print("Median =",np.median(combined))
print("Mode =",mode)
```

```python
A=np.array([[1,-2,3],[-1,3,-1],[2,-5,5]])
B=np.array([9,-6,17])
X1=np.linalg.solve(A,B)
X2=np.dot(np.linalg.inv(A),B)
print("Using linalg.solve")
print(X1)
print("Using inverse matrix")
print(X2)
```

```python
import matplotlib.pyplot as plt
subjects=['Python','DSA','DBMS','OS','CN']
semester1=[82,78,75,80,77]
semester2=[88,84,79,85,83]
plt.figure(figsize=(8,5))
plt.plot(subjects,semester1,marker='o',linewidth=3,color='blue',label='Semester 1')
plt.plot(subjects,semester2,marker='s',linewidth=3,color='green',label='Semester 2')
plt.title('Semester Result Comparison',fontsize=14,color='red')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.grid(True,linestyle='--')
plt.legend()
plt.show()
```
