```python
import pandas as pd
a=pd.DataFrame({'ID':[1,2],'Name':['A','B']})
b=pd.DataFrame({'ID':[3,4],'Name':['C','D']})
c=pd.DataFrame({'ID':[1,2,3,4],'Marks':[91,82,73,64]})
print(a)
print(b)
print(c)
```

```python
v=pd.concat([a,b],ignore_index=True)
print(v)
```

```python
m=pd.merge(v,c,on='ID')
print(m)
```

```python
print('join() works mainly with indexes and is simpler for index-based operations.')
print('merge() is more flexible and works mainly with columns like SQL joins.')
```
