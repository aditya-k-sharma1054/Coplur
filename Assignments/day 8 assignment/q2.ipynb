```python
import pandas as pd
a=pd.DataFrame({'ID':[1,2,3,4],'Name':['A','B','C','D']})
b=pd.DataFrame({'ID':[2,3,4,5],'Marks':[80,90,70,60]})
print(a)
print(b)
```

```python
x=pd.merge(a,b,on='ID',how='inner')
print(x)
```

```python
y=pd.merge(a,b,on='ID',how='left')
print(y)
print('Missing values appear as NaN where no matching ID exists in df2.')
```

```python
r=pd.merge(a,b,on='ID',how='right')
print(r)
```

```python
i1=a.set_index('ID')
i2=b.set_index('ID')
j=i1.join(i2,how='inner')
print(j)
print('merge() joins on columns, while join() joins on index by default.')
```

```python
c=pd.DataFrame({'ID':[1,1,2],'Sec':['X','Y','X'],'Val':[10,20,30]})
d=pd.DataFrame({'ID':[1,2,2],'Sec':['X','X','Y'],'Score':[100,200,300]})
m=pd.merge(c,d,on=['ID','Sec'],how='inner')
print(m)
```
