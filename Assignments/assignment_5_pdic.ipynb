```python
numbers = (12, 5, 27, 8, 19)

total = 0
highest = numbers[0]
lowest = numbers[0]

for num in numbers:
    total += num
    if num > highest:
        highest = num
    if num < lowest:
        lowest = num

print("Total:", total)
print("Highest:", highest)
print("Lowest:", lowest)
```

```python
n = (3, 14, 7, 22, 9, 41, 18, 5)
result = ()
for value in n:
    if value > 10:
        result += (value,)

print("Filtered Tuple:", result)
```

```python
s = {"cat", "dog", "bird", "fish"}
count = 0
for item in s:
    count += 1

print("Number of elements:", count)
```

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}
result = s1.union(s2)

print("Combined Set:", result)
```

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}
common = s1.intersection(s2)

print("Common Elements:", common)
```
