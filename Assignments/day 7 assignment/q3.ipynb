```python
import pandas as pd

tab = pd.DataFrame({
    "Movie": ["Leo", "Jawan", "Animal"],
    "Year": [2023, 2023, 2023]
})

extra = pd.DataFrame([["Stree", 2018]], columns=["Movie", "Year"])

top = tab.iloc[:1]
bottom = tab.iloc[1:]

pd.concat([top, extra, bottom]).reset_index(drop=True)
```
