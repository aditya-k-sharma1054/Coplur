```python
while(True):
    print("""
            1. Add
            2. Remove
            3. Display first e
            4. Display last element
            5. Print
            6. Exit""")
    ch = int(input("Enter your choice: "))  
    if ch == 1:
        e = int(input("Enter element to add: "))
        q.put(e)
    elif ch == 2:
        if not q.empty():
            print(f"Removed element: {q.get()}")
        else:
            print("Queue is empty!")
    elif ch == 3:
        if not q.empty():
            print(f"First element: {q.queue[0]}")
        else:
            print("Queue is empty!")
    elif ch == 4:
        if not q.empty():
            print(f"Last element: {q.queue[-1]}")
        else:
            print("Queue is empty!")
    elif ch == 5:
        print(f"Queue contents: {list(q.queue)}")
    elif ch == 6:
        break
    else:
        print("Invalid choice! Please try again.")
```
