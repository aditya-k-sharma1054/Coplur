```python
# Definitions and rules:
# Fibonacci sequence: each term is the sum of the two previous terms, starting with 0 and 1.
# Armstrong number: a number that is equal to the sum of its own digits each raised to the power of the number of digits.
# Catalan numbers: a combinatorial sequence where C_n = (2n)! / ((n+1)! * n!).
# Triangular numbers: the nth triangular number is n*(n+1)/2, representing dots that form an equilateral triangle.
# Factorial sequence: n! is the product of all positive integers up to n.
# Collatz sequence: for a starting number, if it is even divide by 2, if odd multiply by 3 and add 1, repeat until reaching 1.

def fibonacci_sequence(count: int) -> list[int]:
    seq = [0, 1]
    for _ in range(2, count):
        seq.append(seq[-1] + seq[-2])
    return seq[:count]

def is_armstrong(num: int) -> bool:
    digits = str(abs(num))
    power = len(digits)
    return sum(int(d) ** power for d in digits) == abs(num)

def catalan_numbers(count: int) -> list[int]:
    return [math.comb(2 * i, i) // (i + 1) for i in range(count)]

def triangular_numbers(count: int) -> list[int]:
    return [i * (i + 1) // 2 for i in range(1, count + 1)]

def factorial_sequence(count: int) -> list[int]:
    return [math.factorial(i) for i in range(1, count + 1)]

def collatz_sequence(start: int) -> list[int]:
    seq = [start]
    while seq[-1] != 1:
        n = seq[-1]
        seq.append(n // 2 if n % 2 == 0 else 3 * n + 1)
    return seq

fib_seq_10 = fibonacci_sequence(n)
armstrong_nums_10 = [i for i in range(1, n + 1) if is_armstrong(i)]
catalan_nums_10 = catalan_numbers(n)
triangular_nums_10 = triangular_numbers(n)
factorial_seq_10 = factorial_sequence(n)
collatz_seq_10 = collatz_sequence(n)

print("Fibonacci sequence (n=10):", fib_seq_10)
print("Armstrong numbers up to 10:", armstrong_nums_10)
print("Catalan numbers (first 10):", catalan_nums_10)
print("Triangular numbers (first 10):", triangular_nums_10)
print("Factorial sequence (1 to 10):", factorial_seq_10)
print("Collatz sequence starting from 10:", collatz_seq_10)
```
