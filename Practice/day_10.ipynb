```python
import pandas as pd 
df = pd.read_csv('data.csv') 
print(df)
```

```python
print(df['Name'].value_counts());  print('\n')
print(pd.options.display.max_rows);  print('\n')
print(df.info());                    print('\n')
print(df.describe());                print('\n')
print(df.shape);                     print('\n')
print(df.columns)
```

```python
removed_rows = df[df.isna().any(axis=1)]
print(removed_rows)
```

```python
ndf = df.dropna() 
print(ndf.to_string()) 
df.dropna(inplace = True)
print(df.to_string()) 
```

```python
x = df["mobile"].median() 
df.fillna({"mobile": x}, inplace=True) 
print(x)

x = df["mobile"].mode()[0] 
df.fillna({"mobile": x}, inplace=True) 
print(x)
```
