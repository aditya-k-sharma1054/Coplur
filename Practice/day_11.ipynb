```python
import numpy as np 
# Single random float 
random_float = np.random.rand() 
print("random_float:",random_float) 
# 1D array of random floats 
array_1d = np.random.rand(5) 
print("array_1d:",array_1d) 
# 2D array of random floats 
array_2d = np.random.rand(2, 3) 
print("array_2d:",array_2d) 
# 3D array of random floats 
array_3d = np.random.rand(2, 3, 4) 
print("array_3d:",array_3d) 
```

```python
empty_array = np.empty((2,4),dtype=int,order='F') 
print(empty_array)
```
