```python
import qrcode
def generate_qr_code(data, filename):
    # Configure the QR code parameters
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=10,
        border=4,
    )
    qr.add_data(data)
    qr.make(fit=True)
    
    # Render and save the image
    img = qr.make_image(fill_color="black", back_color="white")
    img.save(filename)

if __name__ == "__main__":
    data = "https://www.example.com"
    filename = "example_qr.png"
    
    # Call the function to actually generate and save the file
    generate_qr_code(data, filename)
    print(f"QR code successfully saved as {filename}")
```

```python
import cv2
def read_qr_code(filename):
    # Load the image
    img = cv2.imread(filename)
    # Initialize the QR code detector
    detector = cv2.QRCodeDetector()
    # Detect and decode the QR code
    data, vertices_array, binary_qrcode = detector.detectAndDecode(img)
    
    if vertices_array is not None:
        print(f"QR Code detected: {data}")
    else:
        print("No QR Code detected.")
if __name__ == "__main__":
    read_qr_code(filename)
```

```python
import numpy as np

def is_valid(grid, row, col, num):
    """Check if placing num in grid[row][col] is valid."""
    # 1. Check the row using NumPy vectorization
    if num in grid[row, :]:
        return False
        
    # 2. Check the column using NumPy vectorization
    if num in grid[:, col]:
        return False
        
    # 3. Check the 3x3 subgrid using NumPy slicing
    start_row, start_col = 3 * (row // 3), 3 * (col // 3)
    subgrid = grid[start_row:start_row+3, start_col:start_col+3]
    if num in subgrid:
        return False
        
    return True

def solve_sudoku(grid):
    """Solve the Sudoku puzzle using backtracking."""
    # Find all empty spaces (where value is 0)
    empty_spaces = np.argwhere(grid == 0)
    
    # If no empty spaces remain, the puzzle is solved
    if len(empty_spaces) == 0:
        return True
        
    # Pick the first empty cell
    row, col = empty_spaces[0]
    
    # Try placing digits from 1 to 9
    for num in range(1, 10):
        if is_valid(grid, row, col, num):
            grid[row, col] = num  # Tentatively place the number
            
            if solve_sudoku(grid):  # Recurse
                return True
                
            grid[row, col] = 0  # Backtrack if the placement fails
            
    return False

# --- Example Usage ---
if __name__ == "__main__":
    # Define an unsolved Sudoku puzzle (0 represents empty cells)
    puzzle = np.array([
        [5, 3, 0, 0, 7, 0, 0, 0, 0],
        [6, 0, 0, 1, 9, 5, 0, 0, 0],
        [0, 9, 8, 0, 0, 0, 0, 6, 0],
        [8, 0, 0, 0, 6, 0, 0, 0, 3],
        [4, 0, 0, 8, 0, 3, 0, 0, 1],
        [7, 0, 0, 0, 2, 0, 0, 0, 6],
        [0, 6, 0, 0, 0, 0, 2, 8, 0],
        [0, 0, 0, 4, 1, 9, 0, 0, 5],
        [0, 0, 0, 0, 8, 0, 0, 7, 9]
    ])

    print("Original Sudoku Puzzle:")
    print(puzzle)
    print("\nSolving...\n")

    if solve_sudoku(puzzle):
        print("Solved Sudoku Grid:")
        print(puzzle)
    else:
        print("No solution exists for this Sudoku puzzle.")
```
