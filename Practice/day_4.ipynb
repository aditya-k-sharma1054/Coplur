```python
# U S E R - D E F I N E D - f u n c t i o n s 
def goo(a,b,/): # (,/) To hide variable name for position-only args
    print(a,b)
goo(1,2)

def foo(*,c,d):# (*,) To show variable names for key-only args
    print(c,d)
foo(c=3,d=4)
```
