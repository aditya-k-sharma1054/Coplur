```python
import pandas as pd
df1 = pd.DataFrame({'Id':[1,2,3,4], 'Name':['A1','A2','A3','A4'], 'Subject':['sub1','sub2','sub3','sub4']})
df2 = pd.DataFrame({'Id':[2,1,3,4], 'Name':['B1','B2','B3','B4'], 'Subject':['sub5','sub1','sub4','sub6']})                    
print(df1);                        print("\n")
print(df2);                        print("\n")
print(pd.concat([df1, df2]));      print("\n")
print("\n Addition\n", df1+df2);   print("\n")
                          
```

```python
# R O W - W I S E - C O N C A T E N A T I O N :
pd.concat([df1, df2], axis=1)
#pd.concat([df1, df2], axis=0)
```

```python
# Syntax for joining on a specific column
pd.merge(df1, df2, on='Id', how='left')
```

```python
print(pd.merge(df1, df2, left_on='Name', right_on='Subject', how='outer'))
```

```python
print(df1.join(df2, lsuffix='_left', rsuffix='_right'))
```

```python
res = pd.concat([df1,df2], keys=['x', 'y'])
print(res)
res.loc['x']
```

```python
ws = df1[df1["Name"] != 0]
print(ws)
```

```python
import pandas as pd
# Inner join: keeps only rows where 'A' matches in both (1 and 2)
rs = pd.merge(df1, df2, on='Subject', suffixes=('_df1', '_df2'))
print(rs)
```

```python
# Left Join: Keeps all rows from df1
pd.merge(df1, df2, on='Subject', how='left')
```

```python
# Right Join: Keeps all rows from df2
pd.merge(df1, df2, on='Subject', how='right')
```

```python
# Outer Join: Keeps all rows from both (fills blanks with NaN)
pd.merge(df1, df2, on='Subject', how='outer')
```

```python
res = pd.merge(df1, df2, left_index=True, right_index=True, suffixes=('_1', '_2'))
print(res)

res = pd.concat([df1, df2], ignore_index=True)
print(res)
```

```python
res = df1.merge(df2,on='Subject',how='left')
print(res)
```

```python
res = df1.merge(df2,on='Subject',how='right')
print(res)
```

```python
res = df1.merge(df2,on='Subject',how='outer')
print(res)
```

```python
res = df1.merge(df2,on='Subject',how='inner')
print(res)
```
